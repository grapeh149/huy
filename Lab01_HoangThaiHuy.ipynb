# 3.2: Prime Numbers

```python
import math
numbers = range(0, 101)
for i in numbers:
    if i < 2:
        print(f"{i}: NO")
        continue
    is_prime = True
    for j in range(2, int(math.sqrt(i)) + 1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{i}: YES")
    else:
        print(f"{i}: NO")
```

# 4.1: Calculator Function

```python
def sum (a, b):
    return a + b
def subtract (a, b):
    return a - b
def multiply (a, b): 
    return a * b
def divide (a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
        return None
print(f"Divide: {divide(0, 0)}")
print(f"Sum: {sum(1, 2)}")
print(f"Subtract: {subtract(0, -5)}")
print(f"Multiply: {multiply(5, 10)}")
```

# 4.2: Recursive Function

```python
def sumStat (a):
    result = 0
    for i in a:
        result += i
    return result
    
def statistics(nums):
    total = sumStat(nums)
    avg = total / len(nums)
    return total, avg, max(nums), min(nums), max(nums)

data = [1, 2, 3, 4, 5]
print(statistics(data))
```

# OOP: Employee Class & Manager Class

```python
class Employee:
    def __init__ (self, name, salary):
        self.name = name
        self.salary = salary
    @property
    def salary(self):
        return self._salary
    @salary.setter
    def salary(self, value):
        if value > 0:
            self._salary = value
        else:
            raise ValueError("Positive only!")

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus
    def total_income(self):
        return self.salary + self.bonus     

m = Manager("Huy", 1000, 500)
print(m.name)           # Huy
print(m.salary)         # 1000
print(m.total_income()) # 1500
```

# Bank System

```python
class Account:
    def __init__(self, money):
        self.money = money
    @property
    def money(self):
        return self._money
    @money.setter
    def money(self, value):
        if value >= 0:
            self._money = value
        else:
            raise ValueError("Positive only!")
    def withdraw(self, amount):
        if amount <= 0:
            print("So tien rut khong hop le")
        elif amount > self.money:
            print(f"Tai khoan khong du de rut so tien {amount}")
        else:
            self.money -= amount
            print(f"Da rut thanh cong so tien {amount}")
    def deposit(self, amount):
        if amount > 0:
            self.money += amount
            print("Gui tien thanh cong")
        else:
            print("So tien gui khong hop le")

x = Account(1000)
x.deposit(1)
x.withdraw(200)
print(f"So du hien tai: {x.money}")
```
