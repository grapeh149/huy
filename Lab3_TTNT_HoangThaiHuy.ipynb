## Exercise 5.1: Basic Plots

```python
# Create the following plots:
# 1. Line plot of y = x^2 for x in [-5, 5]
# 2. Scatter plot with 100 random points, color by quadrant
# 3. Bar chart of 5 products and their sales
# 4. Histogram of 1000 normally distributed values

import numpy as np
import matplotlib.pyplot as plt

# 1. Line plot of y = x^2 for x in [-5, 5]
x = np.linspace(-5, 5, 100)
y = x**2

plt.figure()
plt.plot(x, y)
plt.title("Line Plot: y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.show()


# 2. Scatter plot with 100 random points, color by quadrant
np.random.seed(0)
x_rand = np.random.uniform(-10, 10, 100)
y_rand = np.random.uniform(-10, 10, 100)

# Assign quadrant labels
quadrants = [(x_rand[i] >= 0, y_rand[i] >= 0) for i in range(100)]
quad_labels = [int(q[0]) * 2 + int(q[1]) for q in quadrants]

plt.figure()
plt.scatter(x_rand, y_rand, c=quad_labels)
plt.title("Scatter Plot by Quadrant")
plt.xlabel("x")
plt.ylabel("y")
plt.show()


# 3. Bar chart of 5 products and their sales
products = ["A", "B", "C", "D", "E"]
sales = [150, 200, 300, 250, 180]

plt.figure()
plt.bar(products, sales)
plt.title("Bar Chart: Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()


# 4. Histogram of 1000 normally distributed values
data = np.random.normal(0, 1, 1000)

plt.figure()
plt.hist(data, bins=30)
plt.title("Histogram of Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Exercise 5.2: Customization

```python
# Create a publication-quality figure:
# - Two subplots (side by side)
# - Left: line plot with grid, legend, labels
# - Right: bar chart with annotations
# - Add overall title
# - Save as PNG (300 DPI) and PDF

import numpy as np
import matplotlib.pyplot as plt

# Data for line plot
x = np.linspace(-5, 5, 100)
y = x**2

# Data for bar chart
products = ["A", "B", "C", "D", "E"]
sales = [150, 200, 300, 250, 180]

# Create figure with 2 subplots (side by side)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# -----------------------
# Left: Line plot
# -----------------------
axes[0].plot(x, y, label="y = x^2")
axes[0].set_title("Line Plot")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].grid(True)
axes[0].legend()

# -----------------------
# Right: Bar chart
# -----------------------
bars = axes[1].bar(products, sales)
axes[1].set_title("Product Sales")
axes[1].set_xlabel("Products")
axes[1].set_ylabel("Sales")

# Add annotations on bars
for bar in bars:
    height = bar.get_height()
    axes[1].annotate(f'{height}',
                     xy=(bar.get_x() + bar.get_width() / 2, height),
                     xytext=(0, 3),  # offset
                     textcoords="offset points",
                     ha='center', va='bottom')

# -----------------------
# Overall title
# -----------------------
fig.suptitle("Publication-Quality Figure Example", fontsize=14)

# Adjust layout
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Save figure
plt.savefig("figure.png", dpi=300)
plt.savefig("figure.pdf")

# Show plot
plt.show()
```

## Exercise 5.3: Data Visualization

```python
# Using Titanic dataset:
# Create a 2x2 subplot figure:
# 1. Bar chart of survival by class
# 2. Histogram of ages
# 3. Pie chart of gender distribution
# 4. Scatter plot of age vs fare

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load Titanic dataset
df = sns.load_dataset("titanic")

# Create 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# -----------------------
# 1. Bar chart: Survival by class
# -----------------------
survival_by_class = df.groupby("class")["survived"].mean()
axes[0, 0].bar(survival_by_class.index.astype(str), survival_by_class.values)
axes[0, 0].set_title("Survival Rate by Class")
axes[0, 0].set_xlabel("Class")
axes[0, 0].set_ylabel("Survival Rate")

# -----------------------
# 2. Histogram of ages
# -----------------------
axes[0, 1].hist(df["age"].dropna(), bins=30)
axes[0, 1].set_title("Age Distribution")
axes[0, 1].set_xlabel("Age")
axes[0, 1].set_ylabel("Frequency")

# -----------------------
# 3. Pie chart of gender distribution
# -----------------------
gender_counts = df["sex"].value_counts()
axes[1, 0].pie(gender_counts.values, labels=gender_counts.index, autopct="%1.1f%%")
axes[1, 0].set_title("Gender Distribution")

# -----------------------
# 4. Scatter plot: Age vs Fare
# -----------------------
axes[1, 1].scatter(df["age"], df["fare"])
axes[1, 1].set_title("Age vs Fare")
axes[1, 1].set_xlabel("Age")
axes[1, 1].set_ylabel("Fare")

# -----------------------
# Layout
# -----------------------
plt.tight_layout()
plt.show()
```

# Seaborn

## Exercise 6.1: Distribution Analysis

```python
# Using tips dataset:
# 1. Create histogram of total_bill with KDE
# 2. Box plot of tip by day and time
# 3. Violin plot of total_bill by sex
# 4. Compare distributions of tips for lunch vs dinner

# Style
sns.set_theme(style="whitegrid")

# Load dataset
df = sns.load_dataset("tips")

# Create 2x2 subplot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# -----------------------
# 1. Histogram of total_bill with KDE
# -----------------------
sns.histplot(df["total_bill"], bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Distribution of Total Bill")
axes[0, 0].set_xlabel("Total Bill")
axes[0, 0].set_ylabel("Frequency")

# -----------------------
# 2. Box plot of tip by day and time
# -----------------------
sns.boxplot(x="day", y="tip", hue="time", data=df, ax=axes[0, 1])
axes[0, 1].set_title("Tip Distribution by Day and Time")
axes[0, 1].set_xlabel("Day")
axes[0, 1].set_ylabel("Tip")
axes[0, 1].legend(title="Time")

# -----------------------
# 3. Violin plot of total_bill by sex
# -----------------------
sns.violinplot(x="sex", y="total_bill", data=df, ax=axes[1, 0])
axes[1, 0].set_title("Total Bill Distribution by Sex")
axes[1, 0].set_xlabel("Sex")
axes[1, 0].set_ylabel("Total Bill")

# -----------------------
# 4. Compare distributions of tips (Lunch vs Dinner)
# -----------------------
sns.kdeplot(
    data=df,
    x="tip",
    hue="time",
    fill=True,
    common_norm=False,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Tip Distribution: Lunch vs Dinner")
axes[1, 1].set_xlabel("Tip")
axes[1, 1].set_ylabel("Density")

# -----------------------
# Layout
# -----------------------
fig.suptitle("Tips Dataset Analysis (Advanced Visualization)", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Save
plt.savefig("tips_advanced.png", dpi=300)
plt.savefig("tips_advanced.pdf")

plt.show()
```

## Exercise 6.2: Categorical Analysis


```python
# Using titanic dataset:
# 1. Count plot of survival by class
# 2. Bar plot of average fare by class and sex
# 3. Point plot of survival rate by age group and class
# 4. Create faceted plot by embark_town

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Style
sns.set_theme(style="whitegrid")

# Load dataset
df = sns.load_dataset("titanic")

# -----------------------
# Feature Engineering: Age groups
# -----------------------
bins = [0, 12, 18, 35, 60, 100]
labels = ["Child", "Teen", "Young Adult", "Adult", "Senior"]
df["age_group"] = pd.cut(df["age"], bins=bins, labels=labels)

# -----------------------
# Create 2x2 subplot
# -----------------------
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# -----------------------
# 1. Count plot of survival by class
# -----------------------
sns.countplot(x="class", hue="survived", data=df, ax=axes[0, 0])
axes[0, 0].set_title("Survival Count by Class")
axes[0, 0].set_xlabel("Class")
axes[0, 0].set_ylabel("Count")
axes[0, 0].legend(title="Survived", labels=["No", "Yes"])

# -----------------------
# 2. Bar plot: average fare by class and sex
# -----------------------
sns.barplot(x="class", y="fare", hue="sex", data=df, ax=axes[0, 1])
axes[0, 1].set_title("Average Fare by Class and Sex")
axes[0, 1].set_xlabel("Class")
axes[0, 1].set_ylabel("Average Fare")
axes[0, 1].legend(title="Sex")

# -----------------------
# 3. Point plot: survival rate by age group and class
# -----------------------
sns.pointplot(
    x="age_group",
    y="survived",
    hue="class",
    data=df,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Survival Rate by Age Group and Class")
axes[1, 0].set_xlabel("Age Group")
axes[1, 0].set_ylabel("Survival Rate")
axes[1, 0].legend(title="Class")

# -----------------------
# 4. Faceted plot by embark_town
# -----------------------
g = sns.catplot(
    x="class",
    y="survived",
    hue="sex",
    col="embark_town",
    kind="bar",
    data=df
)

g.fig.suptitle("Survival Rate by Class, Sex, and Embark Town", y=1.05)

# -----------------------
# Layout
# -----------------------
fig.suptitle("Titanic Dataset Analysis (Advanced Multi-Plot)", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```

## Exercise 6.3: Relationship Analysis


```python
# Using iris dataset:
# 1. Scatter plot of sepal_length vs sepal_width by species
# 2. Pair plot of all features
# 3. Heatmap of correlations
# 4. Joint plot of petal dimensions

# Style
sns.set_theme(style="whitegrid")

# Load dataset
df = sns.load_dataset("iris")

# -----------------------
# 1. Scatter plot: sepal_length vs sepal_width by species
# -----------------------
plt.figure(figsize=(6, 5))
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="sepal_width",
    hue="species"
)
plt.title("Sepal Length vs Width by Species")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.tight_layout()
plt.show()

# -----------------------
# 2. Pair plot of all features
# -----------------------
sns.pairplot(df, hue="species")
plt.suptitle("Pair Plot of Iris Features", y=1.02)
plt.show()

# -----------------------
# 3. Heatmap of correlations
# -----------------------
plt.figure(figsize=(6, 5))
corr = df.drop(columns=["species"]).corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()

# -----------------------
# 4. Joint plot of petal dimensions
# -----------------------
sns.jointplot(
    data=df,
    x="petal_length",
    y="petal_width",
    kind="reg"
)
plt.suptitle("Petal Length vs Width (Joint Plot)", y=1.02)
plt.show()
```

## Exercise 6.4: Complete Dashboard

```python
# Create a comprehensive analysis figure:
# - 2x3 subplot grid
# - Include: histogram, box plot, bar chart,
# scatter plot, heatmap, violin plot
# - Use consistent color scheme
# - Add proper titles and labels

# -----------------------
# Style & color palette
# -----------------------
sns.set_theme(style="whitegrid")
palette = "deep"  # consistent color scheme

# Load dataset
df = sns.load_dataset("tips")

# Create 2x3 subplot grid
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# -----------------------
# 1. Histogram
# -----------------------
sns.histplot(df["total_bill"], bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Distribution of Total Bill")
axes[0, 0].set_xlabel("Total Bill")
axes[0, 0].set_ylabel("Frequency")

# -----------------------
# 2. Box Plot
# -----------------------
sns.boxplot(x="day", y="total_bill", data=df, palette=palette, ax=axes[0, 1])
axes[0, 1].set_title("Total Bill by Day")
axes[0, 1].set_xlabel("Day")
axes[0, 1].set_ylabel("Total Bill")

# -----------------------
# 3. Bar Chart
# -----------------------
sns.barplot(x="day", y="tip", data=df, palette=palette, ax=axes[0, 2])
axes[0, 2].set_title("Average Tip by Day")
axes[0, 2].set_xlabel("Day")
axes[0, 2].set_ylabel("Average Tip")

# -----------------------
# 4. Scatter Plot
# -----------------------
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="time",
    palette=palette,
    data=df,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Total Bill vs Tip")
axes[1, 0].set_xlabel("Total Bill")
axes[1, 0].set_ylabel("Tip")

# -----------------------
# 5. Heatmap
# -----------------------
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", ax=axes[1, 1])
axes[1, 1].set_title("Correlation Heatmap")

# -----------------------
# 6. Violin Plot
# -----------------------
sns.violinplot(x="day", y="tip", data=df, palette=palette, ax=axes[1, 2])
axes[1, 2].set_title("Tip Distribution by Day")
axes[1, 2].set_xlabel("Day")
axes[1, 2].set_ylabel("Tip")

# -----------------------
# Layout
# -----------------------
fig.suptitle("Comprehensive Data Analysis Dashboard", fontsize=18)

plt.tight_layout(rect=[0, 0, 1, 0.96])

# Save high-quality outputs
plt.savefig("comprehensive_analysis.png", dpi=300)
plt.savefig("comprehensive_analysis.pdf")

plt.show()
```
