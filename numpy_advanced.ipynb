## Exercise 2.1: Broadcasting Practice

```python
# Create a 5x5 distance matrix where element [i,j]
# is the absolute difference |i - j|
# Expected:
# [[0 1 2 3 4]
# [1 0 1 2 3]
# [2 1 0 1 2]
# [3 2 1 0 1]
# [4 3 2 1 0]]
import numpy as np

n = 5

# Create array [0,1,2,3,4]
arr = np.arange(n)

# Compute distance matrix
dist_matrix = np.abs(arr[:, None] - arr)

print(dist_matrix)
```

## Exercise 2.2: Statistics

```python
# Given: 100 random exam scores (0-100)
scores = np.random.randint(0, 101, 100)
# Calculate:
# 1. Mean, median, std
# 2. Number of students who passed (≥50)
# 3. Percentage of A grades (≥90)
# 4. Grade distribution (A, B, C, D, F)

import numpy as np

np.random.seed(42)

# Given data
scores = np.random.randint(0, 101, 100)

# 1. Mean, median, std
mean_score = np.mean(scores)
median_score = np.median(scores)
std_score = np.std(scores)

print("Mean:", mean_score)
print("Median:", median_score)
print("Std Dev:", std_score)

# 2. Number of students who passed (≥50)
passed = np.sum(scores >= 50)
print("Passed:", passed)

# 3. Percentage of A grades (≥90)
a_percentage = np.mean(scores >= 90) * 100
print("A grade %:", a_percentage)

# 4. Grade distribution
grades = {
    "A": np.sum(scores >= 90),
    "B": np.sum((scores >= 80) & (scores < 90)),
    "C": np.sum((scores >= 70) & (scores < 80)),
    "D": np.sum((scores >= 60) & (scores < 70)),
    "F": np.sum(scores < 60)
}

print("Grade distribution:", grades)
```

## Exercise 2.3: Linear Algebra

```python
# 1. Solve the system:
# x + 2y + 3z = 14
# 2x + 5y + 3z = 18
# 4x + 6y + 8z = 32
# 2. Find eigenvalues of:
# [[4, -2],
# [1, 1]]

import numpy as np

# Coefficient matrix A
A = np.array([
    [1, 2, 3],
    [2, 5, 3],
    [4, 6, 8]
])

# Right-hand side vector b
b = np.array([14, 18, 32])

# Solve Ax = b
solution = np.linalg.solve(A, b)

print("Solution (x, y, z):", solution)

# Matrix
M = np.array([
    [4, -2],
    [1, 1]
])

# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(M)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```
