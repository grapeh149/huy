## Exercise 4.1: Data Cleaning


```python
# Create DataFrame with:
# - 100 rows, columns: ID, Name, Age, Salary, Department
# - Introduce 10% missing values randomly
# - Clean the data:
# 1. Fill Age with median
# 2. Fill Salary with mean by Department
# 3. Fill Name with 'Unknown'
# 4. Drop rows where all values are NaN

import pandas as pd
import numpy as np

# 1. Create DataFrame
np.random.seed(42)

n = 100
df = pd.DataFrame({
    "ID": range(1, n + 1),
    "Name": np.random.choice(["Alice", "Bob", "Charlie", "David", "Emma"], n),
    "Age": np.random.randint(20, 60, n),
    "Salary": np.random.randint(30000, 100000, n),
    "Department": np.random.choice(["HR", "IT", "Finance", "Marketing"], n)
})

# 2. Introduce 10% missing values randomly
for col in df.columns:
    df.loc[df.sample(frac=0.1).index, col] = np.nan

# 3. Cleaning

# Fill Name with 'Unknown'
df["Name"] = df["Name"].fillna("Unknown")

# Fill Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Salary with mean by Department
df["Salary"] = df.groupby("Department")["Salary"].transform(
    lambda x: x.fillna(x.mean())
)

# Drop rows where all values are NaN
df = df.dropna(how="all")

# Final result
print(df.head())
```

## Exercise 4.3: Merge DataFrames


```python
# Create 3 DataFrames:
# - Orders (order_id, customer_id, product_id, quantity)
# - Customers (customer_id, name, city)
# - Products (product_id, product_name, price)
# Merge to create complete order details with:
# customer name, product name, total amount
import pandas as pd

# Orders
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4],
    "customer_id": [101, 102, 101, 103],
    "product_id": [1001, 1002, 1003, 1001],
    "quantity": [2, 1, 5, 3]
})

# Customers
customers = pd.DataFrame({
    "customer_id": [101, 102, 103],
    "name": ["Alice", "Bob", "Charlie"],
    "city": ["HCM", "Hanoi", "Danang"]
})

# Products
products = pd.DataFrame({
    "product_id": [1001, 1002, 1003],
    "product_name": ["Laptop", "Phone", "Tablet"],
    "price": [1000, 500, 300]
})

# Merge Orders + Customers
df = pd.merge(orders, customers, on="customer_id", how="left")

# Merge with Products
df = pd.merge(df, products, on="product_id", how="left")

# Calculate total amount
df["total_amount"] = df["quantity"] * df["price"]

# FInal DataFrame
final_df = df[[
    "order_id",
    "name",
    "product_name",
    "quantity",
    "price",
    "total_amount"
]]

print(final_df)
```

## Exercise 4.4: Reshape Data

```python
# Given wide format exam scores:
# Name, Math, English, Physics, Chemistry
# Convert to long format:
# Name, Subject, Score
# Then pivot to show average by Subject
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Math": [85, 90, 78],
    "English": [88, 76, 85],
    "Physics": [92, 80, 75],
    "Chemistry": [89, 84, 82]
})

print(df)

# Convert to long format
long_df = df.melt(
    id_vars="Name",
    var_name="Subject",
    value_name="Score"
)

print(long_df)

# Pivot to show average by Subject
avg_by_subject = (
    long_df.groupby("Subject")["Score"]
    .mean()
    .reset_index()
)

print(avg_by_subject)
```
